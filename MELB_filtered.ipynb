```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display
from ipywidgets import VBox, HBox, Label

# Cleaned from MELB.ipynb: only foods used by the four current columns.
ingredients = {'120 Banana': {'cal_per_g': 0.875,
                'protein_per_g': 0.0,
                'carbs_per_g': 0.225,
                'fat_per_g': 0.0,
                'fiber_per_g': 0.0263,
                'soluble_frac': 0.35,
                'insoluble_frac': 0.65},
 'Aldi tender': {'cal_per_g': 0.8928571428571429,
                 'protein_per_g': 0.20535714285714285,
                 'carbs_per_g': 0.0,
                 'fat_per_g': 0.008928571428571428,
                 'fiber_per_g': 0.0,
                 'soluble_frac': None,
                 'insoluble_frac': None},
 'Aldi thigh': {'cal_per_g': 1.25,
                'protein_per_g': 0.16964285714285715,
                'carbs_per_g': 0.0,
                'fat_per_g': 0.0625,
                'fiber_per_g': 0.0,
                'soluble_frac': None,
                'insoluble_frac': None},
 'realgood costco': {'cal_per_g': 1.0714285714285714,
                     'protein_per_g': 0.1875,
                     'carbs_per_g': 0.03571428571428571,
                     'fat_per_g': 0.044642857142857144,
                     'fiber_per_g': 0.0,
                     'soluble_frac': None,
                     'insoluble_frac': None},
 'realgood nug': {'cal_per_g': 1.1607142857142858,
                  'protein_per_g': 0.20535714285714285,
                  'carbs_per_g': 0.03571428571428571,
                  'fat_per_g': 0.017857142857142856,
                  'fiber_per_g': 0.0,
                  'soluble_frac': None,
                  'insoluble_frac': None},
 'Costco chipotle': {'cal_per_g': 1.1764705882352942,
                     'protein_per_g': 0.18823529411764706,
                     'carbs_per_g': 0.03529411764705882,
                     'fat_per_g': 0.029411764705882353,
                     'fiber_per_g': 0.0,
                     'soluble_frac': None,
                     'insoluble_frac': None},
 'Enriched Rice': {'cal_per_g': 3.5555555555555554,
                   'protein_per_g': 0.06666666666666667,
                   'carbs_per_g': 0.8,
                   'fat_per_g': 0.0,
                   'fiber_per_g': 0.010416666666666666,
                   'soluble_frac': 0.1,
                   'insoluble_frac': 0.9},
 '90 Kale': {'cal_per_g': 0.470588,
             'protein_per_g': 0.035,
             'carbs_per_g': 0.0941,
             'fat_per_g': 0.0117647,
             'fiber_per_g': 0.05,
             'soluble_frac': 0.25,
             'insoluble_frac': 0.75},
 '60 spinach': {'cal_per_g': 0.23529411764705882,
                'protein_per_g': 0.023529411764705882,
                'carbs_per_g': 0.03529411764705882,
                'fat_per_g': 0,
                'fiber_per_g': 0.022,
                'soluble_frac': 0.25,
                'insoluble_frac': 0.75},
 '170 Lettuce': {'cal_per_g': 0.23529411764705882,
                 'protein_per_g': 0.011764705882352941,
                 'carbs_per_g': 0.03529411764705882,
                 'fat_per_g': 0.0,
                 'fiber_per_g': 0.011764705882352941,
                 'soluble_frac': 0.25,
                 'insoluble_frac': 0.75},
 '50 Egg (large)': {'cal_per_g': 1.4,
                    'protein_per_g': 0.12,
                    'carbs_per_g': 0,
                    'fat_per_g': 0.1,
                    'fiber_per_g': 0.0,
                    'soluble_frac': None,
                    'insoluble_frac': None},
 '300 kirk veg': {'cal_per_g': 0.3763440860215054,
                  'protein_per_g': 0.021505376344086023,
                  'carbs_per_g': 0.07526881720430108,
                  'fat_per_g': 0.0,
                  'fiber_per_g': 0.022,
                  'soluble_frac': 0.3,
                  'insoluble_frac': 0.7},
 '45/s Greek veg': {'cal_per_g': 0.5294117647058824,
                    'protein_per_g': 0.011764705882352941,
                    'carbs_per_g': 0.058823529411764705,
                    'fat_per_g': 0.029411764705882353,
                    'fiber_per_g': 0.023529411764705882,
                    'soluble_frac': None,
                    'insoluble_frac': None},
 '148Russ potato': {'cal_per_g': 0.7432432432432432,
                    'protein_per_g': 0.02027027027027027,
                    'carbs_per_g': 0.16216216216216217,
                    'fat_per_g': 0,
                    'fiber_per_g': 0.013513513513513514,
                    'soluble_frac': 0.2,
                    'insoluble_frac': 0.8},
 'Nuts kirk': {'cal_per_g': 6.428571428571429,
               'protein_per_g': 0.17857142857142858,
               'carbs_per_g': 0.21428571428571427,
               'fat_per_g': 0.5714285714285714,
               'fiber_per_g': 0.125,
               'soluble_frac': 0.2,
               'insoluble_frac': 0.8},
 'Olive oil': {'cal_per_g': 8.4,
               'protein_per_g': 0.0,
               'carbs_per_g': 0,
               'fat_per_g': 0.9333333333333333,
               'fiber_per_g': 0.0,
               'soluble_frac': None,
               'insoluble_frac': None},
 '22/8': {'cal_per_g': 1.4285714285714286,
          'protein_per_g': 0.19642857142857142,
          'carbs_per_g': 0,
          'fat_per_g': 0.07142857142857142,
          'fiber_per_g': 0.0,
          'soluble_frac': None,
          'insoluble_frac': None},
 'Psyllium husk': {'cal_per_g': 3.5,
                   'protein_per_g': 0.0,
                   'carbs_per_g': 0.8,
                   'fat_per_g': 0.0,
                   'fiber_per_g': 0.7,
                   'soluble_frac': 0.8571428571428571,
                   'insoluble_frac': 0.14285714285714285},
 'hersh new': {'cal_per_g': 2.0,
               'protein_per_g': 0.2,
               'carbs_per_g': 0.2,
               'fat_per_g': 0.1,
               'fiber_per_g': 0.4,
               'soluble_frac': None,
               'insoluble_frac': None},
 '2 percent walmart': {'cal_per_g': 0.5416666666666666,
                       'protein_per_g': 0.03333333333333333,
                       'carbs_per_g': 0.05416666666666667,
                       'fat_per_g': 0.020833333333333332,
                       'fiber_per_g': 0.0,
                       'soluble_frac': None,
                       'insoluble_frac': None},
 '1 percent walmart': {'cal_per_g': 0.4583333333333333,
                       'protein_per_g': 0.0375,
                       'carbs_per_g': 0.05416666666666667,
                       'fat_per_g': 0.010416666666666666,
                       'fiber_per_g': 0.0,
                       'soluble_frac': None,
                       'insoluble_frac': None},
 'skim walmart': {'cal_per_g': 0.375,
                  'protein_per_g': 0.0375,
                  'carbs_per_g': 0.05416666666666667,
                  'fat_per_g': 0.0,
                  'fiber_per_g': 0.0,
                  'soluble_frac': None,
                  'insoluble_frac': None},
 'strawberries': {'cal_per_g': 0.35714285714285715,
                  'protein_per_g': 0.0035714285714285713,
                  'carbs_per_g': 0.09285714285714286,
                  'fat_per_g': 0.0,
                  'fiber_per_g': 0.02142857142857143,
                  'soluble_frac': 0.33,
                  'insoluble_frac': 0.67},
 'straw+pine': {'cal_per_g': 0.5,
                'protein_per_g': 0.005714285714285714,
                'carbs_per_g': 0.11428571428571428,
                'fat_per_g': 0.0,
                'fiber_per_g': 0.014285714285714285,
                'soluble_frac': 0.35,
                'insoluble_frac': 0.65},
 'Cal': {'cal_per_g': 1,
         'protein_per_g': 0,
         'carbs_per_g': 0,
         'fat_per_g': 0,
         'fiber_per_g': 0.0,
         'soluble_frac': None,
         'insoluble_frac': None},
 'Cal2': {'cal_per_g': 1,
          'protein_per_g': 0,
          'carbs_per_g': 0,
          'fat_per_g': 0,
          'fiber_per_g': 0.0,
          'soluble_frac': None,
          'insoluble_frac': None},
 'Cal3': {'cal_per_g': 1,
          'protein_per_g': 0,
          'carbs_per_g': 0,
          'fat_per_g': 0,
          'fiber_per_g': 0.0,
          'soluble_frac': None,
          'insoluble_frac': None},
 'Cal4': {'cal_per_g': 1,
          'protein_per_g': 0,
          'carbs_per_g': 0,
          'fat_per_g': 0,
          'fiber_per_g': 0.0,
          'soluble_frac': None,
          'insoluble_frac': None},
 'Protein': {'cal_per_g': 0,
             'protein_per_g': 1,
             'carbs_per_g': 0,
             'fat_per_g': 0,
             'fiber_per_g': 0.0,
             'soluble_frac': None,
             'insoluble_frac': None},
 'carb': {'cal_per_g': 0,
          'protein_per_g': 0,
          'carbs_per_g': 1,
          'fat_per_g': 0,
          'fiber_per_g': 0.0,
          'soluble_frac': None,
          'insoluble_frac': None},
 'Fat': {'cal_per_g': 0,
         'protein_per_g': 0,
         'carbs_per_g': 0,
         'fat_per_g': 1,
         'fiber_per_g': 0.0,
         'soluble_frac': None,
         'insoluble_frac': None},
 'JBareLightlyStrips': {'cal_per_g': 1.8279569892473118,
                        'protein_per_g': 0.1827956989247312,
                        'carbs_per_g': 0.11827956989247312,
                        'fat_per_g': 0.06451612903225806,
                        'fiber_per_g': 0.005376344086021506,
                        'soluble_frac': 0.35,
                        'insoluble_frac': 0.65},
 'Aldi Almonds': {'cal_per_g': 6.071428571428571,
                  'protein_per_g': 0.21428571428571427,
                  'carbs_per_g': 0.17857142857142858,
                  'fat_per_g': 0.5714285714285714,
                  'fiber_per_g': 0.10714285714285714,
                  'soluble_frac': 0.2,
                  'insoluble_frac': 0.8},
 'Hash browns': {'cal_per_g': 0.7058823529411765,
                 'protein_per_g': 0.023529411764705882,
                 'carbs_per_g': 0.16470588235294117,
                 'fat_per_g': 0.0,
                 'fiber_per_g': 0.023529411764705882,
                 'soluble_frac': None,
                 'insoluble_frac': None},
 'Aldi salmon': {'cal_per_g': 0.8928571428571429,
                 'protein_per_g': 0.17857142857142858,
                 'carbs_per_g': 0.0,
                 'fat_per_g': 0.026785714285714284,
                 'fiber_per_g': 0.0,
                 'soluble_frac': None,
                 'insoluble_frac': None}}

sliders = {
    name: widgets.IntSlider(
        value=0, min=0, max=1000, step=10,
        description=name,
        layout=widgets.Layout(width='300px')
    )
    for name in ingredients
}

def calculate_macros(**kwargs):
    totals = {"CalcCalories": 0.0, "Calories": 0.0, "Protein": 0.0, "Carbs": 0.0, "Fat": 0.0}
    fiber_tot = soluble_tot = insoluble_tot = unknown_breakdown = 0.0

    for item, grams in kwargs.items():
        meta = ingredients[item]
        ppg = meta["protein_per_g"]
        cg = meta["carbs_per_g"]
        fg = meta["fat_per_g"]
        kcal_g = meta["cal_per_g"]
        fpg = meta.get("fiber_per_g", 0.0)
        s_frac = meta.get("soluble_frac")
        i_frac = meta.get("insoluble_frac")

        totals["CalcCalories"] += grams * (ppg*4 + cg*4 + fg*9)
        totals["Calories"] += grams * kcal_g
        totals["Protein"] += grams * ppg
        totals["Carbs"] += grams * cg
        totals["Fat"] += grams * fg

        item_fiber = grams * float(fpg)
        fiber_tot += item_fiber
        if s_frac is not None and i_frac is not None:
            soluble_tot += item_fiber * s_frac
            insoluble_tot += item_fiber * i_frac
        else:
            unknown_breakdown += item_fiber

    display(pd.DataFrame(totals, index=["Total"]).round(2))

    ratio = round(soluble_tot / insoluble_tot, 2) if insoluble_tot > 0 else ""
    display(pd.DataFrame({
        "Total Fiber (31g)": [round(fiber_tot, 2)],
        "Insoluble Fiber (g)": [round(insoluble_tot, 2)],
        "Soluble Fiber (g)": [round(soluble_tot, 2)],
        "Soluble:Insoluble (.33)": [ratio],
    }, index=["Total"]))

MELB_sources = ['120 Banana', 'Aldi tender', 'Aldi thigh', 'realgood costco', 'realgood nug', 'Costco chipotle', 'Enriched Rice', '90 Kale', '60 spinach', '170 Lettuce', '50 Egg (large)', '300 kirk veg', '45/s Greek veg', '148Russ potato', 'Nuts kirk', 'Olive oil', '22/8', 'Psyllium husk']
cream_sources = ['hersh new', '2 percent walmart', '1 percent walmart', 'skim walmart', 'strawberries', 'straw+pine']
things = ['Cal', 'Cal2', 'Cal3', 'Cal4', 'Protein', 'carb', 'Fat']
test_sources = ['JBareLightlyStrips', 'Aldi Almonds', 'Hash browns', 'Aldi salmon']

column1 = VBox([Label("Meal")] + [sliders[i] for i in MELB_sources])
column2 = VBox([Label("Cream")] + [sliders[i] for i in cream_sources])
column3 = VBox([Label("Calories")] + [sliders[i] for i in things])
column4 = VBox([Label("Test")] + [sliders[i] for i in test_sources])

display(HBox([column1, column2, column3, column4]))
widgets.interactive_output(calculate_macros, sliders)
```
